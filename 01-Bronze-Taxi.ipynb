```python
%python
import dlt
import re
from pyspark.sql.functions import *

TAXI_SOURCE_TABLE = "/Volumes/students_data/team-1-data-schema/team-1-data-schema-volume/taxi_data.csv"
```

```python
%sql
USE CATALOG `students_data`;
USE SCHEMA `team-1-data-schema`;
```

```python
%python
@dlt.table(
    name="bronze_taxi_data",
    comment="Raw taxi data ingested from source csv file"
)
def bronze_taxi_data():
    """
    This function defines our 'bronze_taxi_data' table.
    The table name is set in the @dlt.table decorator.
    """
    df = spark.read.csv(TAXI_SOURCE_TABLE, header=True)
    df = df.withColumn("source_file", col("_metadata.file_path"))
    df = df.toDF(*[c.replace(" ", "_").lower() for c in df.columns])
    return df.withColumn("ingestion_timestamp", current_timestamp())
```
